```python
def run_simulator(graph, start, goal, actions):

    current = start

    path = [start]

    print("Starting location:", start)

    print("Target location:", goal)

    for action in actions:

        if current == goal:

            break

        if action not in graph.get(current, []):

            print("Invalid move:", current, "to", action)

            return {

                "path": path,

                "current": current,

                "reached_goal": False

            }

        current = action

        path.append(current)

    reached = current == goal

    result = {

        "path": path,

        "current": current,

        "reached_goal": reached

    }

    print("Path:", path)

    print("Final location:", current)

    print("Goal reached:", reached)

    return result

# Test the simulator

test_actions = [assigned["graph"][assigned["start"]][0]]

simulation_result = run_simulator(

    assigned["graph"],

    assigned["start"],

    assigned["goal"],

    test_actions

)

```

## 5. Uninformed search: BFS

Implement BFS using a FIFO queue, repeated-state tracking and route reconstruction. Explain why it is uninformed and why it finds a minimum-move path on these unit-cost graphs. Show the route returned for your assigned mission and the agent executing its service action. Handle start=goal, no route and invalid locations correctly. No separate test-case table or results-analysis task is required.

```python
from collections import deque

def bfs(graph, start, goal):

    queue = deque([[start]])

    visited = {start}

    while queue:

        path = queue.popleft()

        current = path[-1]

        if current == goal:

            return path

        for neighbor in graph[current]:

            if neighbor not in visited:

                visited.add(neighbor)

                new_path = path + [neighbor]

                queue.append(new_path)

    return None

# Run BFS on your assigned mission

bfs_path = bfs(

    assigned["graph"],

    assigned["start"],

    assigned["goal"]

)

print("Start:", assigned["start"])

print("Goal:", assigned["goal"])

print("BFS path:", bfs_path)

if bfs_path:

    print("Number of moves:", len(bfs_path) - 1)

else:

    print("No path found.")
```

## 6. Informed search: A*

Implement A* with f(n) = g(n) + h(n), using a priority queue, best-cost tracking and parent reconstruction. Use the supplied HEURISTICS table for your assigned goal. Explain what g and h mean, why the supplied h is admissible on the provided variant, and when it would need revising. Test the goal on removal from the frontier. Show the assigned route and execute the service action. A* and BFS should find the same minimum move count here; their paths may differ. No separate comparison table or limitations essay is required.

```python
import heapq
def heuristic(node, goal):
   return HEURISTICS[goal][node]
def a_star(graph, start, goal):
  if start not in graph or goal not in graph:
    return None
  priority_queue = [(heuristic(start, goal), 0, start)]

  best_cost = {start: 0}
  parent = {}

  while priority_queue:
    f_cost, g_cost, current = heapq.heappop(priority_queue)

    if g_cost != best_cost.get(current):
        continue

    if current == goal:
        path = []
        node = goal

        while node in parent:
            path.append(node)
            node = parent[node]

        path.append(start)
        return path[::-1]

    for neighbor in graph[current]:
        new_cost = g_cost + 1

        if new_cost < best_cost.get(neighbor, float("inf")):
            best_cost[neighbor] = new_cost
            parent[neighbor] = current

            f = new_cost + heuristic(neighbor, goal)

            heapq.heappush(
                priority_queue,
                (f, new_cost, neighbor)
            )


  return None
astar_path = a_star(
    assigned["graph"],
    assigned["start"],
    assigned["goal"]
    )
print("Start:", assigned["start"])
print("Goal:", assigned["goal"])
print("A* path:", astar_path)
if astar_path:
  print("Number of moves:", len(astar_path) - 1)
else: print("No path found.")
```
